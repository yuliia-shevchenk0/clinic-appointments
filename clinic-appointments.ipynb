```python
import numpy as np 
import pandas as pd 
```

```python
#suppress the use of scientific notation, set the maximum character width per line, limit the number of digits 2 decimal places
np.set_printoptions(suppress = True, linewidth=100, precision=2)
```

```python
#DF with messy data of clinic appointments 
raw_clinic_data = pd.read_csv('clinic-appointments.csv')
raw_clinic_data
```

```python
#making copy of original df
df = raw_clinic_data.copy()
```

```python
#info about df: rows, NaN and non-null cells, datatypes
df.info()
```

```python
#checing for duplicates
df.duplicated().sum()
```

```python
#checing for duplicates in "patient_name" column
d = {x : y.iloc[:, :6]  for x, y in df.groupby(['patient_name']) if len(y) > 1}
d
```

```python
#age checking
```

```python
max(df['age'])
```

```python
min(df['age'])
```

```python
#writing patients names in alphabetical order
df = df.sort_values('patient_name')
df
```

```python
#reset index and drop the old index
df = df.reset_index(drop=True)
df
```

```python
#change patients ID
```

```python
#set new IDs, from 1000 to 1999
df['patient_id'] = range(1000, 1000 + len(df))
df
```

```python
#checkpoint 1. names and IDs 
df_names = df
df_checkpoint = df_names.copy()
df_checkpoint
```

```python
#inconsistent naming of genders
```

```python
df_names['gender'].unique()
```

```python
df_names['gender'] = df_names['gender'].replace({'f': 'F', 'female': 'F', 'Female': 'F', '0': 'F', 'm': 'M', 'male': 'M', 'Male': 'M', '1': 'M'})
df_names['gender'] = df_names['gender'].fillna('X')
df_names['gender'].unique()
```

```python
#inconsistent answers to "follow_up_required"
```

```python
df_names['follow_up_required'].unique()
```

```python
df_names['follow_up_required'] = df_names['follow_up_required'].replace({'No': '0', 'N': '0', 'Yes': '1', 'Y': '1'})
df_names['follow_up_required'].unique()
```

```python
#checkpoint 2. genders and answers
df_gen_an = df_names
df_checkpoint = df_gen_an.copy()
df_checkpoint
```

```python
#inconsistent date types in appointment_date and booking_date
```

```python
df_gen_an['appointment_date']
```

```python
df_gen_an['appointment_date'] = pd.to_datetime(df_gen_an['appointment_date'], format='mixed')
df_gen_an['appointment_date']
```

```python
df_gen_an['booking_date']
```

```python
df_gen_an['booking_date'] = pd.to_datetime(df_gen_an['booking_date'], format='mixed')
df_gen_an['booking_date']
```

```python
#checking the dates logic
invalid_dates = df_gen_an[df_gen_an['appointment_date'] < df_gen_an['booking_date']]
invalid_dates
```

```python
#checkpoint 3. dates
df_dates = df_gen_an
df_checkpoint = df_dates.copy()
df_checkpoint
```

```python
#convert different currencies to $
```

```python
#checking for nan values
pd.isna(df_dates['billing_amount']).sum()
```

```python
#convert different currencies to $ depending on currency, ignoring the nan values
pd.options.display.float_format = '{:.2f}'.format
money = df_dates['billing_amount'].astype(str)

conversion_map = { #14.05.2026 
    "€": "1.17*", 
    "Rs": "0.01*", 
    "\$": "1*", 
    "£": "1.35*"
}

df_dates['billing_usd'] = money.replace(conversion_map, regex=True).map(pd.eval, na_action='ignore')
df_dates['billing_usd']
```

```python
#saving original billing_amount
billing_amount = df_dates['billing_amount']
```

```python
#dropping original billing_amount
df_dates = df_dates.drop(columns=['billing_amount'])
df_dates
```

```python
#filling nan values with average value of each department
df_dates["billing_usd"] = df_dates.groupby("department")["billing_usd"].transform(lambda x: x.fillna(x.mean()))
df_dates["billing_usd"]
```

```python
#checkpoint 4. money
df_money = df_dates
df_checkpoint = df_money.copy()
df_checkpoint
```

```python
#creating new column for calculating days of waiting between booking and appointment
df_money["waiting_days"] = (
    df_money["appointment_date"] - df_money["booking_date"]
).dt.days
df_money
```

```python
#creating new column for age groups
df_money['age_group'] = pd.cut(df_money['age'], bins = [18,35,50,70,90], labels = ['Young adult', 'Adult', 'Middle age', 'Senior'])
df_money['age_group']
```

```python
#reordering columns
```

```python
df_money.columns.values
```

```python
column_names_reordered = ['patient_id','patient_name','age','age_group', 'gender','appointment_date','booking_date', 'waiting_days', 'doctor','department', 'billing_usd', 'follow_up_required',]
```

```python
df_money = df_money[column_names_reordered]
df_money
```

```python
#checkpoint 5. clean dataset
df_clean = df_money
df_checkpoint = df_clean.copy()
df_checkpoint
```

```python
df_clean.to_csv('clinic-appointments-clean.csv')
```
